```python
# 🔗 Logic: Lấy món ăn dựa trên quốc gia yêu thích và tính tỷ lệ
print("🔗 LOGIC MAPPING: QUỐC GIA YÊU THÍCH → MÓN ĂN TƯƠNG ỨNG + TỶ LỆ")
print("="*80)

# Mapping giữa quốc gia yêu thích và cột món ăn tương ứng
COUNTRY_TO_DISH_COLUMN = {
    # Vietnamese
    'Vietnamese': 'Bạn thích những món ăn Việt Nam nào sau đây? (Which Vietnamese dishes do you like?)',
    'Việt Nam': 'Bạn thích những món ăn Việt Nam nào sau đây? (Which Vietnamese dishes do you like?)',
    'Việt Nam (Vietnamese)': 'Bạn thích những món ăn Việt Nam nào sau đây? (Which Vietnamese dishes do you like?)',
    
    # Thai
    'Thai': 'Bạn thích những món ăn Thái Lan nào sau đây? (Which Thai dishes do you like?)',
    'Thái Lan': 'Bạn thích những món ăn Thái Lan nào sau đây? (Which Thai dishes do you like?)',
    'Thái Lan (Thai)': 'Bạn thích những món ăn Thái Lan nào sau đây? (Which Thai dishes do you like?)',
    
    # Chinese
    'Chinese': 'Bạn thích những món ăn Trung Quốc nào sau đây? (Which Chinese dishes do you like?)  ',
    'Trung Quốc': 'Bạn thích những món ăn Trung Quốc nào sau đây? (Which Chinese dishes do you like?)  ',
    'Trung Quốc (Chinese)': 'Bạn thích những món ăn Trung Quốc nào sau đây? (Which Chinese dishes do you like?)  ',
    
    # French
    'French': 'Bạn thích những món ăn Pháp nào sau đây? (Which French dishes do you like?)',
    'Pháp': 'Bạn thích những món ăn Pháp nào sau đây? (Which French dishes do you like?)',
    'Pháp (French)': 'Bạn thích những món ăn Pháp nào sau đây? (Which French dishes do you like?)',
    
    # Indian
    'Indian': 'Bạn thích những món ăn Ấn Độ nào sau đây? (Which Indian dishes do you like?)  ',
    'Ấn Độ': 'Bạn thích những món ăn Ấn Độ nào sau đây? (Which Indian dishes do you like?)  ',
    'Ấn Độ (Indian)': 'Bạn thích những món ăn Ấn Độ nào sau đây? (Which Indian dishes do you like?)  ',
    
    # Italian
    'Italian': 'Bạn thích những món ăn Ý nào sau đây? (Which Italian dishes do you like?)  ',
    'Ý': 'Bạn thích những món ăn Ý nào sau đây? (Which Italian dishes do you like?)  ',
    'Ý (Italian)': 'Bạn thích những món ăn Ý nào sau đây? (Which Italian dishes do you like?)  ',
    
    # Japanese
    'Japanese': 'Bạn thích những món ăn Nhật Bản nào sau đây? (Which Japanese dishes do you like?)',
    'Nhật Bản': 'Bạn thích những món ăn Nhật Bản nào sau đây? (Which Japanese dishes do you like?)',
    'Nhật Bản (Japanese)': 'Bạn thích những món ăn Nhật Bản nào sau đây? (Which Japanese dishes do you like?)',
    
    # Korean
    'Korean': 'Bạn thích những món ăn Hàn Quốc nào sau đây? (Which Korean dishes do you like?)',
    'Hàn Quốc': 'Bạn thích những món ăn Hàn Quốc nào sau đây? (Which Korean dishes do you like?)',
    'Hàn Quốc (Korean)': 'Bạn thích những món ăn Hàn Quốc nào sau đây? (Which Korean dishes do you like?)',
    
    # American
    'American': 'Bạn thích những món ăn Mỹ? (Which American dishes do you like?)',
    'Mỹ': 'Bạn thích những món ăn Mỹ? (Which American dishes do you like?)',
    'Mỹ (American)': 'Bạn thích những món ăn Mỹ? (Which American dishes do you like?)',
}

# Tìm tên cột quốc gia từ DataFrame (tránh lỗi ký tự đặc biệt)
country_col = None
for col in df.columns:
    if 'quốc gia' in col.lower() and 'cuisine' in col.lower():
        country_col = col
        break

if country_col is None:
    # Fallback: tìm cột có chứa "quốc gia"
    matching_cols = [col for col in df.columns if 'quốc gia' in col.lower()]
    if matching_cols:
        country_col = matching_cols[0]

print(f"✅ Đã tìm thấy cột: '{country_col}'")

# Hàm phân tích tỷ lệ món ăn cho một quốc gia
def analyze_dish_preferences_by_country(country_name, country_keywords):
    """
    Phân tích tỷ lệ món ăn yêu thích cho một quốc gia cụ thể
    
    Args:
        country_name: Tên quốc gia để hiển thị (ví dụ: "Việt Nam")
        country_keywords: List các từ khóa để tìm (ví dụ: ["Vietnamese", "Việt Nam"])
    
    Returns:
        DataFrame với thống kê món ăn và tỷ lệ
    """
    if country_col not in df.columns:
        return None
    
    # Tìm các user chọn quốc gia này
    pattern = '|'.join(country_keywords)
    users_with_country = df[df[country_col].str.contains(pattern, case=False, na=False)]
    total_users = len(users_with_country)
    
    if total_users == 0:
        print(f"   ❌ Không tìm thấy user nào yêu thích {country_name}")
        return None
    
    print(f"\n📍 {country_name.upper()}:")
    print(f"   - Tổng số user chọn quốc gia này: {total_users}")
    
    # Lấy cột món ăn tương ứng - ưu tiên tìm trực tiếp từ tên quốc gia
    dish_col = None
    
    # Cách 1: Tìm trực tiếp từ DataFrame dựa trên tên quốc gia (chính xác nhất)
    country_name_lower = country_name.lower()
    for col in df.columns:
        if 'món ăn' in col.lower() and country_name_lower in col.lower():
            dish_col = col
            print(f"   ✅ Tìm thấy cột món ăn trực tiếp: '{dish_col[:60]}...'")
            break
    
    # Cách 2: Nếu không tìm thấy, dùng mapping dictionary
    if not dish_col or dish_col not in df.columns:
        for keyword in country_keywords:
            dish_col = COUNTRY_TO_DISH_COLUMN.get(keyword)
            if dish_col and dish_col in df.columns:
                print(f"   ✅ Tìm thấy cột món ăn từ mapping (keyword '{keyword}'): '{dish_col[:60]}...'")
                break
    
    # Xác minh lại cột có đúng không
    if dish_col and dish_col in df.columns:
        # Kiểm tra tên cột có chứa tên quốc gia không
        if country_name_lower not in dish_col.lower():
            print(f"   ⚠️ CẢNH BÁO: Cột món ăn có vẻ không khớp với '{country_name}'!")
            print(f"   - Tên cột: '{dish_col}'")
            # Thử tìm lại
            for col in df.columns:
                if 'món ăn' in col.lower() and country_name_lower in col.lower():
                    dish_col = col
                    print(f"   ✅ Đã sửa lại cột đúng: '{dish_col[:60]}...'")
                    break
    
    if not dish_col or dish_col not in df.columns:
        print(f"   ❌ Không tìm thấy cột món ăn tương ứng cho '{country_name}'")
        print(f"   🔍 Đã thử keywords: {country_keywords}")
        return None
    
    # Lấy món ăn của các user chọn quốc gia này
    dishes_data = users_with_country[dish_col].dropna()
    users_with_dishes = len(dishes_data)
    
    print(f"   - Số user có dữ liệu món ăn: {users_with_dishes}/{total_users} ({users_with_dishes/total_users*100:.1f}%)")
    
    # Phân tích từng món ăn
    from collections import Counter
    all_dishes = []
    
    for dish_str in dishes_data:
        if pd.notna(dish_str):
            # Tách các món ăn (phân cách bằng dấu phẩy)
            dishes = [d.strip() for d in str(dish_str).split(',') if d.strip()]
            all_dishes.extend(dishes)
    
    # Đếm số lần xuất hiện của mỗi món
    dish_counts = Counter(all_dishes)
    
    # Tạo DataFrame kết quả
    results = []
    for dish, count in dish_counts.most_common():
        percentage = (count / total_users) * 100
        results.append({
            'Món ăn (Dish)': dish,
            'Số user thích (Users)': count,
            'Tổng user chọn quốc gia (Total)': total_users,
            'Tỷ lệ (%)': f"{percentage:.2f}%",
            'Tỷ lệ (số)': percentage
        })
    
    result_df = pd.DataFrame(results)
    
    print(f"\n   📊 TOP 10 MÓN ĂN PHỔ BIẾN NHẤT:")
    print("   " + "-" * 70)
    for idx, row in result_df.head(10).iterrows():
        print(f"   {idx+1:2d}. {row['Món ăn (Dish)']:<40} | {row['Số user thích (Users)']:>3}/{row['Tổng user chọn quốc gia (Total)']:<3} | {row['Tỷ lệ (%)']:>7}")
    
    return result_df

# Kiểm tra và hiển thị mẫu giá trị
if country_col and country_col in df.columns:
    print(f"\n📋 Mẫu giá trị từ cột '{country_col}':")
    sample_values = df[country_col].head(3)
    for idx, val in sample_values.items():
        print(f"  - {val}")
else:
    print(f"\n❌ Không tìm thấy cột quốc gia!")

# Ví dụ 1: Phân tích món ăn Việt Nam
print("\n" + "="*80)
print("📌 VÍ DỤ 1: PHÂN TÍCH MÓN ĂN VIỆT NAM")
print("="*80)

vietnamese_stats = analyze_dish_preferences_by_country(
    "Việt Nam", 
    ["Vietnamese", "Việt Nam"]
)

# Hiển thị kết quả nếu có
if vietnamese_stats is not None:
    print(f"\n✅ Phân tích thành công! Tổng số món ăn: {len(vietnamese_stats)}")
    print("\n📊 DataFrame kết quả:")
    print(vietnamese_stats.head(10))
else:
    print("\n❌ Không có kết quả trả về. Kiểm tra lại logic.")
```

```python
# 📊 Tạo bảng tổng hợp: Quốc gia → Số lượng user → Món ăn phổ biến
print("📊 BẢNG TỔNG HỢP: QUỐC GIA YÊU THÍCH VÀ MÓN ĂN TƯƠNG ỨNG")
print("="*80)

country_col = ' Bạn yêu thích món ăn của quốc gia nào?'

if country_col in df.columns:
    summary_data = []
    
    # Danh sách các quốc gia và cột món ăn tương ứng
    countries_mapping = {
        'Vietnamese': {
            'name_vi': 'Việt Nam',
            'dish_col': 'Bạn thích những món ăn Việt Nam nào sau đây? (Which Vietnamese dishes do you like?)'
        },
        'Thai': {
            'name_vi': 'Thái Lan',
            'dish_col': 'Bạn thích những món ăn Thái Lan nào sau đây? (Which Thai dishes do you like?)'
        },
        'Chinese': {
            'name_vi': 'Trung Quốc',
            'dish_col': 'Bạn thích những món ăn Trung Quốc nào sau đây? (Which Chinese dishes do you like?)  '
        },
        'French': {
            'name_vi': 'Pháp',
            'dish_col': 'Bạn thích những món ăn Pháp nào sau đây? (Which French dishes do you like?)'
        },
        'Indian': {
            'name_vi': 'Ấn Độ',
            'dish_col': 'Bạn thích những món ăn Ấn Độ nào sau đây? (Which Indian dishes do you like?)  '
        },
        'Italian': {
            'name_vi': 'Ý',
            'dish_col': 'Bạn thích những món ăn Ý nào sau đây? (Which Italian dishes do you like?)  '
        },
        'Japanese': {
            'name_vi': 'Nhật Bản',
            'dish_col': 'Bạn thích những món ăn Nhật Bản nào sau đây? (Which Japanese dishes do you like?)'
        },
        'Korean': {
            'name_vi': 'Hàn Quốc',
            'dish_col': 'Bạn thích những món ăn Hàn Quốc nào sau đây? (Which Korean dishes do you like?)'
        },
        'American': {
            'name_vi': 'Mỹ',
            'dish_col': 'Bạn thích những món ăn Mỹ? (Which American dishes do you like?)'
        }
    }
    
    for country_key, info in countries_mapping.items():
        dish_col = info['dish_col']
        name_vi = info['name_vi']
        
        # Đếm số user chọn quốc gia này
        users_with_country = df[df[country_col].str.contains(country_key, case=False, na=False)]
        num_users = len(users_with_country)
        
        # Đếm số user có món ăn tương ứng
        if dish_col in df.columns:
            users_with_dishes = df[dish_col].notna().sum()
            
            # Lấy món ăn của những user chọn quốc gia này
            if num_users > 0:
                dishes_list = users_with_country[dish_col].dropna().tolist()
                # Đếm món ăn phổ biến (có thể có nhiều món trong một chuỗi)
                all_dishes = []
                for dish_str in dishes_list:
                    if pd.notna(dish_str):
                        # Tách các món ăn (phân cách bằng dấu phẩy)
                        dishes = [d.strip() for d in str(dish_str).split(',')]
                        all_dishes.extend(dishes)
                
                from collections import Counter
                dish_counts = Counter(all_dishes)
                top_dishes = dish_counts.most_common(3)
                
                summary_data.append({
                    'Quốc gia (Country)': f"{name_vi} ({country_key})",
                    'Số user chọn (Users)': num_users,
                    'Số user có món ăn (Users with dishes)': users_with_dishes,
                    'Món ăn phổ biến nhất (Top dishes)': ', '.join([f"{dish} ({count})" for dish, count in top_dishes])
                })
    
    summary_df = pd.DataFrame(summary_data)
    print("\n")
    print(summary_df.to_string(index=False))
    
    print("\n" + "="*80)
    print("✅ LOGIC HOÀN CHỈNH:")
    print("   - Nếu user chọn quốc gia yêu thích → Lấy món ăn từ cột tương ứng")
    print("   - Ví dụ: User chọn 'Vietnamese' → Lấy từ cột 'Bạn thích những món ăn Việt Nam nào sau đây?'")
```
